# AI vs Machine Learning vs Deep Learning

This notebook compares Artificial Intelligence (AI), Machine Learning (ML) and Deep Learning (DL) using definitions, a comparison table, real-world examples and an explanation of how they are related.

## 1. Definitions

**Artificial Intelligence (AI)** is the broad field of building machines that can perform tasks that normally need human intelligence, such as reasoning, problem solving, understanding language and recognising images.

**Machine Learning (ML)** is a subset of AI. Instead of being given fixed rules, an ML system learns patterns from data and uses them to make predictions or decisions.

**Deep Learning (DL)** is a subset of ML that uses neural networks with many layers. It learns useful features from raw data such as images, audio and text on its own.

## 2. How they are related

The three ideas are nested inside each other like circles:

    +--------------------------------------------------+
    |  Artificial Intelligence (AI)                    |
    |   +------------------------------------------+   |
    |   |  Machine Learning (ML)                   |   |
    |   |   +----------------------------------+   |   |
    |   |   |  Deep Learning (DL)              |   |   |
    |   |   +----------------------------------+   |   |
    |   +------------------------------------------+   |
    +--------------------------------------------------+

- **AI** is the broader field. Its goal is intelligent behaviour.
- **ML** is one way to achieve AI: let the machine learn from data.
- **DL** is one way to do ML: use deep neural networks.

Every deep learning system is a machine learning system, and every machine learning system is an AI system, but not the other way round. A simple rule-based chess program is AI but not ML.

```python
fields = {"AI": {"AI"}, "ML": {"AI", "ML"}, "DL": {"AI", "ML", "DL"}}

for name, belongs_to in fields.items():
    print(f"{name} is part of: {', '.join(sorted(belongs_to - {name})) or 'nothing (it is the broadest field)'}")
```

```python
examples = {
    "AI (rule-based)": ["Chess program using fixed rules", "Customer support bot with scripted answers", "Automatic traffic signal timing"],
    "Machine Learning": ["Email spam filter", "Movie and product recommendation systems", "Predicting house prices from past sales"],
    "Deep Learning": ["Face and image recognition", "Voice assistants (speech recognition)", "Language translation and chatbots"],
}

for field, items in examples.items():
    print(field)
    for item in items:
        print("  -", item)
    print()
```

## 3. Summary

| | AI | ML | DL |
|---|---|---|---|
| **One line** | Make machines intelligent | Let machines learn from data | Learn with deep neural networks |
| **Relation** | The whole field | Inside AI | Inside ML |
| **Example** | Chess rules | Spam filter | Image recognition |

**Key takeaway:** AI is the goal, Machine Learning is a method to reach it, and Deep Learning is a powerful kind of Machine Learning for large and complex data.
